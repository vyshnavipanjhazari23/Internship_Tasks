```python
# Q1. Find the shortest word

# Observe: Input is a sentence.
# Understand: Find the word with the fewest characters.
# Break: Split sentence, compare word lengths.
# Logic: Keep track of the shortest word.
```

```python
sentence = "Python makes coding easy"
words = sentence.split()

shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word

print(shortest)
```

```python
#Q2. Find the Largest Number in a List
#Start with the first number as the largest. 
#Compare every number with it and update the largest value whenever a bigger number appears.
```

```python
numbers = [12, 45, 7, 89, 23]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print(largest)
```

```python
#Q3. Find the SmallestNumber in a List
#Start with the first number as the smallest. 
#If another number is smaller, update the variable.
```

```python
numbers = [12, 45, 7, 89, 23]
smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print(smallest)
```

```python
#Q4. Find the First Longest Word When Ties Occur
#Start with the first word. 
#Update the answer only when a word is strictly longer (>), not when its length is equal.
#This preserves the first word in a tie.
```

```python
sentence = "Python makes coding useful"
words = sentence.split()

longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word

print(longest)
```

```python
#Q5. Find the Word with the Maximum Number of Characters
#We compare the lengths of the words and retain the first word with the maximum length.
```

```python
#Q6. Count Vowels in a String
#Visit each character and count it if it is one of a, e, i, o, u
```

```python
text = "Education"
vowels = "aeiou"
count = 0

for char in text.lower():
    if char in vowels:
        count += 1

print(count)
```

```python
#Q7. Count Consonants in a String
#Count alphabetic characters that are not vowels. Ignore spaces, numbers, and special characters.
```

```python
text = "Python 123!"
vowels = "aeiou"
count = 0

for char in text.lower():
    if char.isalpha() and char not in vowels:
        count += 1

print(count)
```

```python
#Q8. Find the Most Frequent Character
#Ignore spaces, count each character, and select the character with the highest frequency. 
#If frequencies tie, choose the one appearing first.
```

```python
text = "banana".replace(" ", "")
result = text[0]
highest = 0

for char in text:
    frequency = text.count(char)

    if frequency > highest:
        highest = frequency
        result = char

print(result)
```

```python
#Q9. Find the Least Frequent Character
#Count each character and find the smallest frequency. 
#Update only when the frequency is strictly smaller, preserving the first character when there is a tie.
```

```python
text = "success".replace(" ", "")
result = text[0]
lowest = text.count(result)

for char in text:
    frequency = text.count(char)

    if frequency < lowest:
        lowest = frequency
        result = char

print(result)
```

```python
#Q10. Find All Characters Appearing More Than Once
#Check every character's frequency. 
#Add a character to the result only if it occurs more than once and has not already been added.
```

```python
text = "programming"
repeated = []

for char in text:
    if text.count(char) > 1 and char not in repeated:
        repeated.append(char)

print(repeated)
```

```python
#Q14. Dictionary of Numbers and Their Squares
#For every number, create a dictionary entry where the key is the number and the value is number ** 2.
```

```python
#Q15. Create a Dictionary of Numbers and Their Even/Odd Status
#A number is even if number % 2 == 0; otherwise, it is odd.
```

```python
numbers = [1, 2, 3, 4, 5]
result = {}

for number in numbers:
    if number % 2 == 0:
        result[number] = "Even"
    else:
        result[number] = "Odd"

print(result)
```

```python
#Q16. Find the Most Frequent Word in a Sentence
#Convert all words to lowercase so Python and python are treated as the same word. 
#Count each word and preserve the first word when frequencies tie.
```

```python
sentence = "Python is easy and Python is powerful"
words = sentence.lower().split()

result = words[0]
highest = 0

for word in words:
    frequency = words.count(word)

    if frequency > highest:
        highest = frequency
        result = word

print(result)
```

```python
#Q17. Find All Words Appearing More Than Once
#Convert words to lowercase. 
#Add a word to the result if its frequency is greater than one and it has not already been added.
```

```python
sentence = "data science uses data and science tools"
words = sentence.lower().split()
repeated = []

for word in words:
    if words.count(word) > 1 and word not in repeated:
        repeated.append(word)

print(repeated)
```

```python
#Q18. Find the First Word That Appears Only Once
#Examine the words in their original order. The first word whose frequency is exactly one is the answer.
```

```python
sentence = "apple banana apple mango banana orange"
words = sentence.lower().split()

for word in words:
    if words.count(word) == 1:
        print(word)
        break
```
