```python
#A shallow copy creates a new outer list, but its inner lists are shared.
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

```python
#A deep copy duplicates the nested lists, so changing b does not affect a.
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

```

```python
#a.copy() is shallow, so the inner list modified through a is also referenced by b
```

```python
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
```

```python
#Deep copying creates independent nested lists.
```

```python
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
```

```python
#The outer lists are different objects, but their first inner list is shared.
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
```

```python
#*args collects positional arguments into a tuple.
```

```python
def test(*args):
    print(args)

test(1, 2)
```

```python
#The args parameter is always a tuple inside the function.
```

```python
def test(*args):
    print(type(args))

test("a", "b", "c")
```

```python
#**kwargs collects keyword arguments into a dictionary.
```

```python
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
```

```python
#The kwargs parameter is a dictionary inside the function.
```

```python
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)
```
